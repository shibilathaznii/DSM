```python
import matplotlib.pyplot as plt
x = [1, 2, 6, 18]
y = [3, 10, 12, 20]
plt.plot(x,y,linestyle=':', marker='o', color='red')

plt.title(" Line plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
bar_width=0.2
men = [20, 23, 36, 25, 30]
women = [21, 24, 26, 28, 35]
groups = ['G1','G2','G3','G4','G5']
x = np.arange(len(men))
plt.bar(x-bar_width/2, men,bar_width,label="Men")
plt.bar(x+bar_width/2,women,bar_width,label="Women")
plt.ylabel("Scores")
plt.legend()
plt.title("Gender group")
plt.show()
```
